3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

B = np.array([[7,  8,  9],
              [10, 11, 12]])

result_matrix = B @ A
print("result_matrix (2x2):", result_matrix)

C = np.array([[1],
              [2]])

result_C = result_matrix @ C
print("result_matrix @ C (2x1):", result_C)

D = np.array([[1, 0],
              [0, 1]])  # одинична матриця для прикладу

result_D = result_matrix @ D
print("result_matrix @ D (2x2):", result_D)
```
