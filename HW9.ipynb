```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('tip.csv')
```

```python
df['total_bill'].hist(bins=20)
plt.title ('bill', fontsize=20)
plt.xlabel ('total', fontsize=10)
plt.ylabel ('amount', fontsize=10)
```

```python
df.boxplot(column='tip', by='day')
plt.title('Розподіл чайових по дням тижня')
plt.suptitle('')
plt.xlabel ('день тижня', fontsize=10)
plt.ylabel ('чайові', fontsize=10)
print(df.groupby('day')['tip'].median())
```

Найвище медіанне значення чайових у неділю

```python
plt.scatter(df['total_bill'], df['tip'])
```

```python
average_bill=df.groupby('time')['total_bill'].mean()
plt.bar(average_bill.index, average_bill.values)
plt.title ('Середній рахунок за часом дня')
plt.xlabel ('час дня', fontsize=10)
plt.ylabel ('середній рахунок', fontsize=10)
```

```python
import plotly.express as px
import plotly.graph_objects as go
fig=px.scatter (df, x='total_bill', y='tip', color='sex', 
                title='Загальний чек vs чайові за статтю',
                labels={
                    'total_bill' : 'Загальний чек',
                    'tip' : 'Чайові',
                    'sex' : 'Стать'})
fig.show()
```

```python
fig=px.box (df, x='tip', y='day', color='sex',
            title='Чайові за різні дні тижня за кольором статті',
            labels={
                'tip' : 'Чайові',
                'day' : 'День тижня',
                'sex' : 'Стать'
            })
fig.show()
```
