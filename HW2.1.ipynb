Мета: Перевірити, чи середнє значення вибірки відрізняється від гіпотетичного середнього.

Припущення: Середній ріст студентів університету вважається 170 см. Ми хочемо перевірити, чи середній ріст випадкової вибірки студентів істотно більший. Вибірка задана в змінній heights.

Для виконання завдання імпортуйте метод для проведення ztest, виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи справдилось припущення про середній зріст студентів на рівні значущості 0.05?

Використовуємо тут Z-тест, оскільки вибірка достатньо велика (більше 30 екземплярів) і можемо вважати, що її стандартне відхилення близьке до ст. відхилення генеральної сукупності, а значить - відоме.

```python
import numpy as np
from statsmodels.stats.weightstats import ztest
z_stat, p_value = ztest(heights, value=H0_mean, alternative='larger')
```

```python
heights = [174, 171, 175, 179, 170, 170, 179, 175, 169, 174, 169, 169, 173, 162, 163, 169, 166,
           173, 167, 164, 179, 170, 172, 164, 169, 175, 169, 174, 169, 169, 173, 162, 177]
H0_mean= 170
alpha=0.05
```

```python
print(f"Розмір вибірки: n = {len(heights)}")
print(f"Середнє вибірки: {np.mean(heights):.3f}")
print(f"Z-статистика: {z_stat:.3f}")
print(f"p-value: {p_value:.3f}")

if p_value < alpha:
    print(f"p-value ({p_value:.3f}) < alpha ({alpha})")
    print("Відхилити H0: середній зріст студентів істотно більший за 170 см")
else:
    print(f"p-value ({p_value:.3f}) >= alpha ({alpha})")
    print("Не може бути відхилена H0: немає підстав вважати середній зріст істотно більшим за 170 см")
```

Припущення: Група A отримала новий навчальний курс, а група B продовжувала зі стандартним курсом. Ми перевіряємо, чи є новий курс ефективнішим.

Виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи справдилось припущення щодо студентів на рівні значущості 0.05?

```python
group_a_scores = [78.55, 72.25, 79.88, 75., 76.54, 74.99, 87.26, 77.93, 72.71,
                   82.11, 71.9, 79.04, 68.2, 71.36, 78.98, 81.69, 78.86, 77.42,
                   76.49, 70.61, 74.4, 75.7, 83.29, 79.72, 69.18, 79.62, 76.07,
                   74.62, 81.06, 83.15, 82.66, 73.8, 76.45, 79.66, 82.88, 75.6,
                   77.07, 72.47, 72.02, 82.06]

group_b_scores = [81.78, 74.64, 80.02, 76.81, 71.77, 76.81, 82.69, 74.82, 82.82,
                   61.9, 79.11, 75.44, 73.5, 75.46, 65.06, 73.9, 76.79, 82.39,
                   72.41, 70.96, 72.49, 79.58, 76.64, 72.35, 77.57, 75.49, 79.84,
                   71.49, 73.36, 73.04, 67.68, 76.48, 76.31, 75.03, 73.83, 67.92,
                   72.9, 73.29, 70.99, 74.19]

z_stat, p_value = ztest(group_a_scores, group_b_scores, value=0, alternative='two-sided')

alpha = 0.05
print(f"Середнє групи A: {sum(group_a_scores)/len(group_a_scores):.3f}")
print(f"Середнє групи B: {sum(group_b_scores)/len(group_b_scores):.3f}")
print(f"Z-статистика: {z_stat:.3f}")
print(f"p-value: {p_value:.3f}")

if p_value < alpha:
    print(f"p-value ({p_value:.3f}) < alpha ({alpha})")
    print("Відхилити H0: існує статистично значуща різниця між групами, новий курс впливає на результат")
else:
    print(f"p-value ({p_value:.3f}) >= alpha ({alpha})")
    print("Не може бути відхилена H0: немає статистично значущих підстав вважати, що групи відрізняються")
```

Мета: Побачити, як впливає розмір вибірок на результат тесту.

Завдання: Уявіть, що з групи А ми маємо тільки перші 5 записів, а з групи В - усі. Таке могло статись, якщо ми вже маємо результати тестів учнів за попередньою програмою, а тести за новою програмою здали наразі лише 5 учнів і ми вирішили не чекати довше. Виберіть перші 5 записів для групи А і проведіть t-test (для цього використовуємо наступний метод). Виведіть значення p-value. Зробіть висновок, який результат ви бачите щодо стат. значущості на рівні 0.05 для цього експерименту?

Тут проводимо t-test оскільки одна з вибірок - дуже мала

```python
from scipy.stats import ttest_ind
group_a_small = group_a_scores[:5]

t_stat, p_value_ttest = ttest_ind(group_a_small, group_b_scores)

print(f"Перші 5 значень групи A: {group_a_small}")
print(f"Розмір вибірки A: {len(group_a_small)}")
print(f"Розмір вибірки B: {len(group_b_scores)}")
print(f"t-статистика: {t_stat:.3f}")
print(f"p-value: {p_value_ttest:.3f}")

if p_value_ttest < alpha:
    print(f"p-value ({p_value_ttest:.3f}) < alpha ({alpha})")
    print("Відхилити H0: навіть при дуже малій вибірці виявлено статистично значущу різницю")
else:
    print(f"p-value ({p_value_ttest:.3f}) >= alpha ({alpha})")
    print("Не може бути відхилена H0: при такій малій вибірці (n=5) тест не має достатньої статистичної потужності, щоб виявити різницю")
```

При зменишні вибірки та використання т-тесту, статистична потужність сильно падає. при n=5, важще відхилити Н0, тому що довірчий інтервал стає ширшим, і випадкові коливання в малих вибірках мають більший вплив на результат.

Завдання 3*: Z-тест для двох пов'язаних вибірок
Мета: Перевірити, чи вплинув тренінг на продуктивність працівників, коли всі справробітники проходили тренінг.

Припущення: Наші працівники проходять однаковий тренінг, і ми хочемо з'ясувати, чи покращилися їхні показники після тренінгу на рівні значущості 0.05. Виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи поліпшились показники працівників на рівні значущості 0.05?

Зверніть увагу, ці вибірки повʼязані між собою, значить не буде коректно виконати z-test між ними, натомість, ми можемо порівняти, чи різниця в результатах (продуктивності співробітників) є стат значущою.

```python
before_training = [57.82, 37.63, 36.8, 55.22, 52.97, 52.5, 53.46, 43.2, 52.32,
                    52.93, 42.86, 68.66, 54.74, 38.09, 56.57, 40.25, 57.37, 61.59,
                    41.79, 59.63, 54.13, 58.22, 68.97, 47.55, 42.46, 41.1, 41.84,
                    49.23, 53.41, 52.77]

after_training = [62.47, 40.66, 42.7, 57.69, 61.41, 56.76, 54.75, 44.06, 56.29,
                   55.48, 47.25, 72.6, 57.59, 39.39, 56.54, 42.36, 62.58, 65.01,
                   42.3, 62.95, 57.9, 59.45, 72.28, 50.66, 43.18, 44.82, 45.96,
                   54.4, 58.52, 53.01]

differences = [a - b for a, b in zip(after_training, before_training)]

z_stat, p_value = ztest(differences, value=0, alternative='larger')

print(f"Середня різниця (after - before): {sum(differences)/len(differences):.3f}")
print(f"Z-статистика: {z_stat:.3f}")
print(f"p-value: {p_value:.3f}")

if p_value < alpha:
    print(f"p-value ({p_value:.3f}) < alpha ({alpha})")
    print("Відхилити H0: тренінг значно статистично покращив продуктивність працівників")
else:
    print(f"p-value ({p_value:.3f}) >= alpha ({alpha})")
    print("Не може бути відхилена H0: немає вагомих підстав вважати, що тренінг покращив продуктивність")
```
