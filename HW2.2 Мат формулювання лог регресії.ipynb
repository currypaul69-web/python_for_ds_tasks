```python
import numpy as np
```

#1

```python
def sigmoid(z):
    result = 1 / (1 + np.exp(-z))
    return result
z=np.array([-2, -1, 0, 1, 2])

values = sigmoid(z)

print ('Значення сигмоїди')
print ('z | sigmoid(z)')
for i in range(len(z)):
    print(f'{z[i]} | {values[i]:.4f}')
```

#2

```python
def hypothesis(theta, X):
    z = np.dot(X, theta)
    result = sigmoid(z)
    return result

theta = np.array([0.5, -0.5])
X = np.array([
    [1,  2],
    [1, -1],
    [1,  0],
    [1,  1]
])

predictions = hypothesis(theta, X)

print ('Значення гіпотези')
print (f'theta: {theta}')
print (f'X: {X}')
print ()
print ('X       | hypothesis(theta, X)')
for i in range(len(predictions)):
    print(f'{X[i]} | {predictions[i]:.4f}') 
```

#3

```python
def compute_gradient(theta, X, y):
    m = len(y)
    predictions = hypothesis(theta, X)
    errors = predictions - y
    gradient = (1/m) * np.dot(X.T, errors)
    
    return gradient

theta = np.array([0.5, -0.5])

x = np.array([
    [1,  2],
    [1, -1],
    [1,  0],
    [1,  1]
])

y = np.array([1, 0, 0, 1])
grad=compute_gradient(theta, X, y)

print ('Градієнт функції втрат')
print (f'theta: {theta}')
print (f"gradient: {grad}")
```

#4

```python
def full_gradient_descent(theta, X, y, learning_rate=0.1, num_iterations=100):
    m, n = X.shape
    theta = np.zeros(n)

    for epoch in range(num_iterations):
        gradient = compute_gradient(theta, X, y)
        theta = theta - learning_rate * gradient
    return theta

X = np.array([
    [1,  2],
    [1, -1],
    [1,  0],
    [1,  1]
])
y = np.array([1, 0, 0, 1])
theta_final = full_gradient_descent(theta, X, y, learning_rate=0.1, num_iterations=100)
print ('Full batch gradient descent')
print (f'Final theta: {theta_final}')
```

```python
def predict_proba(theta, X):
    z = X @ theta
    proba = 1 / (1 + np.exp(-z))
    return proba

def predict(theta, X, threshold=0.5):
    proba = predict_proba(theta, X)
    return (proba >= threshold).astype(int)

def accuracy(y_true, y_pred):
    correct = (y_true == y_pred)
    return np.sum(correct) / len(y_true)

proba = predict_proba(theta, X)
y_pred = predict(theta, X)
acc = accuracy(y, y_pred)

print("Ймовірності належності до класу 1:")
print(proba)

print("Передбачені класи:")
print(y_pred)

print("Точність моделі:")
print(acc)
```
