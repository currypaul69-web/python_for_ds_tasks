1) Задана функція залежності оцінки від кількості годин, витрачених на навчання:  f(x)=10x−5 .
2) З допомогою Python обчисліть похідну цієї функції.
3) Використовуйте бібліотеку matplotlib для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
f = 10*x - 5
f_prime = sp.diff(f, x)
print("f(x) =", f)
print("f'(x) =", f_prime)

f_num = sp.lambdify(x, f, 'numpy')
f_prime_num = sp.lambdify(x, f_prime, 'numpy')

x_vals = np.linspace(0, 10, 100)
y_vals = f_num(x_vals)
y_prime_vals = np.full_like(x_vals, f_prime_num(0))  # похідна стала

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='f(x)=10x-5')
plt.plot(x_vals, y_prime_vals, label="f'(x)=10", linestyle='--')
plt.xlabel('Кількість годин')
plt.ylabel('Оцінка / Похідна')
plt.title('Залежність оцінки від кількості годин навчання')
plt.legend()
plt.grid(True)
plt.show()
```

Похідна f'(x)=10 — стала, тобто оцінка зростає рівномірно: кожна додаткова година навчання дає +10 балів

Використовуючи Python і SymPy, обчисліть похідні для функцій  f(x)=sin(x)  та  f(x)=ln(x) . Зобразіть ці функції та їхні похідні за допомогою matplotlib.

```python
x = sp.symbols('x')

f1 = sp.sin(x)
f1_prime = sp.diff(f1, x)

f2 = sp.ln(x)
f2_prime = sp.diff(f2, x)

print("f1(x) =", f1, " f1'(x) =", f1_prime)
print("f2(x) =", f2, " f2'(x) =", f2_prime)

f1_num = sp.lambdify(x, f1, 'numpy')
f1_prime_num = sp.lambdify(x, f1_prime, 'numpy')
f2_num = sp.lambdify(x, f2, 'numpy')
f2_prime_num = sp.lambdify(x, f2_prime, 'numpy')

x_vals1 = np.linspace(-2*np.pi, 2*np.pi, 200)
x_vals2 = np.linspace(0.1, 10, 200)

fig, axs = plt.subplots(1, 2, figsize=(10, 5))

axs[0].plot(x_vals1, f1_num(x_vals1), label='sin(x)')
axs[0].plot(x_vals1, f1_prime_num(x_vals1), label="sin'(x)=cos(x)", linestyle='--')
axs[0].set_title('Функція sin(x) та її похідна')
axs[0].legend()
axs[0].grid(True)

axs[1].plot(x_vals2, f2_num(x_vals2), label='ln(x)')
axs[1].plot(x_vals2, f2_prime_num(x_vals2), label="ln'(x)=1/x", linestyle='--')
axs[1].set_title('Функція ln(x) та її похідна')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

Загальні витрати компанії від обсягу виробництва задані функцією  f(x)=x2−5x+1 , де  x  — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x**2-5*x + 1
f_prime = sp.diff(f, x)

print("f(x) =", f)
print("f'(x) =", f_prime)

f_num = sp.lambdify(x, f, 'numpy')
f_prime_num = sp.lambdify(x, f_prime, 'numpy')

x_vals = np.linspace(0, 100, 500)
y_vals = f_num(x_vals)
y_prime_vals = f_prime_num(x_vals)

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='f(x) = x²-5x+1')
plt.plot(x_vals, y_prime_vals, label="f'(x)=2x-5", linestyle='--')
plt.xlabel('Кількість одиниць (x)')
plt.ylabel('Витрати / Похідна')
plt.title('Витрати компанії та їх похідна')
plt.legend()
plt.grid(True)
plt.show()
```
